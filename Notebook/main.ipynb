Handle Missing Values

```python
df['job_skills'].fillna('Not Specified', inplace=True)
df['job_summary'].fillna('Not Available', inplace=True)
df.to_csv("Jobs.csv", index=False)
df.head(3)
```

Check Duplicates Rows

```python
duplicate_rows=df.duplicated().sum()
print(f"Number of Duplicated Rows : {duplicate_rows}")
```

Feature Engineering


```python
df.head(1)
df['first_seen']=pd.to_datetime(df['first_seen'])
df['Year']=df['first_seen'].dt.year
df['Month']=df['first_seen'].dt.month
df['Day']=df['first_seen'].dt.day
df['Month_Name']=df['first_seen'].dt.month_name()
df[['first_seen', 'Year', 'Month', 'Day','Month_Name']].head()
df.to_csv("Jobs.csv", index=False)
```

```python
df.head(3)
```

```python
df['Number_of_Skills']=df['job_skills'].str.split(',').str.len()
df['Number_of_Skills'].head(10)

```

```python
df['Remote_Flag']=df['job_type'].map({'Onsite':0,'Remote':1,'Hybrid':2})
df['Remote_Flag'].head(5)
df.to_csv('Jobs.csv',index=False)
```

Job Title Analysis

```python
# Top 20 most demanded job
df.columns
Top_20_Jobs = (
    df['job_title']
    .value_counts()
    .head(20)
    .reset_index()
)
Top_20_Jobs.columns = ['Job Title', 'Number of Job Postings']
Top_20_Jobs
```

```python
sns.barplot(
    data=Top_20_Jobs,x='Number of Job Postings',y='Job Title',hue='Job Title',palette='crest',      
    dodge=False,
    saturation=0.9,       
    edgecolor='black',
    linewidth=1.5,
    width=0.7,                 
    legend=False
)
plt.title('Top 20 Most Demand Jobs')
plt.xlabel('Number of Job Postings')
plt.ylabel('Jobs')
```

```python
# Least common job
Least_Jobs=df['job_title'].value_counts().tail(10).reset_index(name='Number of Postings')
Least_Jobs
```

```python
sns.barplot(data=Least_Jobs,x='Number of Postings',y='job_title',hue='job_title',palette='Set2',
            edgecolor='black', 
            linewidth=1.5,
            width=0.7, )
plt.title('Least Job Postings')
plt.xlabel('Number of Job Postings')
plt.ylabel('Jobs')
```

```python
data_analyst_jobs = df[df['job_title'] == 'Data Analyst']

Top_Companies = (
    data_analyst_jobs['company']
    .value_counts()
    .head(20)
    .reset_index(name='Number of Job Posting')
)
Top_Companies
```

```python
# Average number of postings per company
avg_posting=df.groupby('company')['job_title'].count().reset_index(name='Number of Postings').sort_values(by='Number of Postings',ascending=False)
avg_posting.head(10)
```

```python
top_10=avg_posting.head(10)
sns.barplot(data=top_10,x='Number of Postings',y='company',hue='company',palette='viridis',edgecolor='black',linewidth=2)
plt.style.use('ggplot')
plt.xlim()
plt.ylim()
plt.title('Average Number of Postings per company')
plt.xlabel('Number of Posting')
plt.ylabel('Companies')
```

```python
companies_multiple_roles = (
    df.groupby("company")["job_title"]
      .nunique()
      .reset_index(name="Number of Job Roles")
      .query("`Number of Job Roles` > 1")
      .sort_values(by="Number of Job Roles", ascending=False)
)
companies_multiple_roles.head(10)
```

Company Analysis

```python
# Companies hiring different experience levels
pivot = pd.pivot_table(
    data=df,
    values='job level',
    index='company',
    aggfunc='nunique'
)

pivot[pivot['job level'] > 1].sort_values(
    by='job level',
    ascending=False
).head(10)
```

```python
# Companies posting highest variety of jobs
pivot = pd.pivot_table(
    data=df,
    values='job_title',
    index='company',
    aggfunc='nunique'
)

pivot.sort_values(by='job_title', ascending=False).head(10)
```

```python
top10=pivot.head(10)
plt.figure(figsize=(12, 6))
sns.barplot(data=top10, x='job_title', y='company',edgecolor='black',linewidth=2,width=0.7)

plt.title('Top 10 Companies Posting the Highest Variety of Jobs')
plt.xlabel('Number of Unique Job Titles')
plt.ylabel('Company')
plt.tight_layout()
plt.show()
```

Location Analysis

```python
# Top countries by job postings

countries_postings = (
    pd.pivot_table(
        data=df,
        values='job_title',
        index='search_country',
        aggfunc='count'
    )
    .reset_index()
    .rename(columns={'job_title': 'Number of Job Posting'})
)

countries_postings.sort_values(
    by='Number of Job Posting',
    ascending=False,
    inplace=True
)

countries_postings
```

```python
sns.barplot(
    data=countries_postings.head(10),
    x='Number of Job Posting',
    y='search_country',
     color='royalblue',
    edgecolor='black',
      
)

plt.title('Top 10 Countries by Job Postings')
plt.xlabel('Number of Job Postings')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

```python
# Top cities
Top_cities=df.groupby('search_city')['job_title'].count().sort_values(ascending=False).reset_index(name='Number of Job Postings')
Top_cities.head(10)
```

```python
sns.barplot(data=Top_cities.head(10),x='Number of Job Postings',y='search_city',edgecolor='black',linewidth=1)
plt.title('Top Cities with Number of Postings')
plt.xlabel('Number of Job Postings')
plt.ylabel('Cities')
plt.tight_layout()
```

```python
# City vs Job Count
job_count = (
    df.groupby('search_city')['job_title']
      .count()
      .reset_index(name='Number of Job Posting')
      .sort_values(by='Number of Job Posting', ascending=False)
)

job_count.head(10)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=job_count.head(10),
    x='Number of Job Posting',
    y='search_city',
    edgecolor='black',
    linewidth=2
)

plt.title('Top 10 Cities by Job Postings')
plt.xlabel('Number of Job Postings')
plt.ylabel('City')
plt.tight_layout()
plt.show()


# Which country has the highest demand?
# As per the given graph Hollywood city have highest demand with Highest Number of Job Openings 
# Which locations have the widest variety of job roles?
# Top10 Cities with highest job posting are shown in horizontal bar chart 
```

Job Level Analysis

```python
# Associate Level jobs
associate_level=df[df['job level']=='Associate']
print(f"Total Associate Level Jobs: {associate_level.shape[0]}")
```

```python
# Mid senior Level Jobs 
mid_senior=df[df['job level']=='Mid senior']
print(f'Total Number of Mid Senior Level Jobs are :{mid_senior.shape[0]}')
```

```python
# Percentage Disturbution
job_level_percentage = (
    df['job level']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
    .reset_index()
)

job_level_percentage.columns = ['Job Level', 'Percentage']
print(job_level_percentage)
```

```python
# Which companies hire mostly mid senior employees
mid_senior=df[df['job level']=='Mid senior']
top_hire=mid_senior.groupby('company').size().reset_index(name='Count of Employee')
top_hire.head(10)
```

```python
sns.barplot(data=top_hire.head(10),x='Count of Employee',y='company',hue='company',palette='crest',edgecolor='black')
plt.title('Top Companies Hiring Mid Senior Employee')
plt.ylabel('Number of Mid Senior Employee')
plt.xlabel('Companies')
plt.tight_layout()
```

Job Type Analysis 

```python
# distribution of job postings across different job types
df.columns
distuburtion_of_job_types=(df['job_type'].value_counts(normalize=True)
    .mul(100)
    .round(2)
    .reset_index()
)
distuburtion_of_job_types.columns=['job_type','percentage']
distuburtion_of_job_types
```

```python
plt.Figure(figsize=(7,7))
plt.pie(distuburtion_of_job_types['percentage'],labels=distuburtion_of_job_types['job_type'],autopct='%1.1f%%',startangle=90,explode=[0.05]*len(distuburtion_of_job_types),   # Slightly separate every slice
    shadow=True,
    wedgeprops={'edgecolor':'white', 'linewidth':2},
    textprops={'fontsize':11})
plt.title('Distribution of Job Postings by Job Type')
plt.tight_layout()
plt.show()
```

```python
# job type has the highest and lowest number of job postings
postings=pd.pivot_table(data=df,values='job_title',index='job_type',aggfunc='count').reset_index().rename(columns={'job_title': 'Number of Job Postings'})
postings.sort_values(by='Number of Job Postings',ascending=False)
```

```python
# companies offer the highest number of Full-time jobs
Onsite=df[df['job_type']=='Onsite']
Onsite.count()
Highest_number_of_jobs=Onsite.groupby('company').size().reset_index(name='Count of Jobs').sort_values(by='Count of Jobs',ascending=False)
Highest_number_of_jobs.head(10)
```

```python
sns.barplot(data=Highest_number_of_jobs.head(10),x='Count of Jobs',y='company',hue='company',palette='muted',edgecolor='black',linewidth=2,width=0.8)
plt.title('companies offer the highest number of Full-time jobs')
plt.xlabel('Count of Full-Time Job Postings')
plt.ylabel('Companies')
plt.tight_layout()
```

```python
# countries have the highest demand for job type
countires=df.groupby('search_country')['job_type'].count().reset_index(name='Count of Job Types').sort_values(by='Count of Job Types',ascending=False)
countires
```

```python
sns.barplot(data=countires,x='search_country',y='Count of Job Types',hue='search_country',palette='colorblind',edgecolor='black',linewidth=1)
plt.title('Countries have the highest demand for job type')
plt.xlabel('Country')
plt.tight_layout()
```

```python
# job type is most common for each experience level
experience_jobtype = pd.pivot_table(
    data=df,
    index='job level',
    columns='job_type',
    values='job_title',
    aggfunc='count',
    fill_value=0
)

experience_jobtype
```

```python
# companies offer the greatest variety of job types
company_job_types = (
    df.groupby('company')['job_type']
      .nunique()
      .sort_values(ascending=False)
      .reset_index(name='Number of Job Types')
)

company_job_types.head(10)
```

```python
top10 = company_job_types.head(10)
plt.figure(figsize=(12,6))
sns.barplot(data=top10,x='Number of Job Types', y='company', hue='company', palette='viridis')
plt.title('Top 10 Companies Offering the Greatest Variety of Job Types',
          fontsize=15, weight='bold')
plt.xlabel('Number of Job Types')
plt.ylabel('Company')
plt.tight_layout()
```
