Handle Missing Values

```python
df['job_skills'].fillna('Not Specified', inplace=True)
df['job_summary'].fillna('Not Available', inplace=True)
df.to_csv("Jobs.csv", index=False)
df.head(3)
```

Check Duplicates Rows

```python
duplicate_rows=df.duplicated().sum()
print(f"Number of Duplicated Rows : {duplicate_rows}")
```

Feature Engineering


```python
df.head(1)
df['first_seen']=pd.to_datetime(df['first_seen'])
df['Year']=df['first_seen'].dt.year
df['Month']=df['first_seen'].dt.month
df['Day']=df['first_seen'].dt.day
df['Month_Name']=df['first_seen'].dt.month_name()
df[['first_seen', 'Year', 'Month', 'Day','Month_Name']].head()
df.to_csv("Jobs.csv", index=False)
```

```python
df.head(3)
```

```python
df['Number_of_Skills']=df['job_skills'].str.split(',').str.len()
df['Number_of_Skills'].head(10)

```

```python
df['Remote_Flag']=df['job_type'].map({'Onsite':0,'Remote':1,'Hybrid':2})
df['Remote_Flag'].head(5)
df.to_csv('Jobs.csv',index=False)
```

Job Title Analysis

```python
# Top 20 most demanded job
df.columns
Top_20_Jobs = (
    df['job_title']
    .value_counts()
    .head(20)
    .reset_index()
)
Top_20_Jobs.columns = ['Job Title', 'Number of Job Postings']
Top_20_Jobs
```

```python
sns.barplot(
    data=Top_20_Jobs,x='Number of Job Postings',y='Job Title',hue='Job Title',palette='crest',      
    dodge=False,
    saturation=0.9,       
    edgecolor='black',
    linewidth=1.5,
    width=0.7,                 
    legend=False
)
plt.title('Top 20 Most Demand Jobs')
plt.xlabel('Number of Job Postings')
plt.ylabel('Jobs')
```

```python
# Least common job
Least_Jobs=df['job_title'].value_counts().tail(10).reset_index(name='Number of Postings')
Least_Jobs
```

```python
sns.barplot(data=Least_Jobs,x='Number of Postings',y='job_title',hue='job_title',palette='Set2',
            edgecolor='black', 
            linewidth=1.5,
            width=0.7, )
plt.title('Least Job Postings')
plt.xlabel('Number of Job Postings')
plt.ylabel('Jobs')
```

```python
data_analyst_jobs = df[df['job_title'] == 'Data Analyst']

Top_Companies = (
    data_analyst_jobs['company']
    .value_counts()
    .head(20)
    .reset_index(name='Number of Job Posting')
)
Top_Companies
```

```python
# Average number of postings per company
avg_posting=df.groupby('company')['job_title'].count().reset_index(name='Number of Postings').sort_values(by='Number of Postings',ascending=False)
avg_posting.head(10)
```

```python
top_10=avg_posting.head(10)
sns.barplot(data=top_10,x='Number of Postings',y='company',hue='company',palette='viridis',edgecolor='black',linewidth=2)
plt.style.use('ggplot')
plt.xlim()
plt.ylim()
plt.title('Average Number of Postings per company')
plt.xlabel('Number of Posting')
plt.ylabel('Companies')
```

```python
companies_multiple_roles = (
    df.groupby("company")["job_title"]
      .nunique()
      .reset_index(name="Number of Job Roles")
      .query("`Number of Job Roles` > 1")
      .sort_values(by="Number of Job Roles", ascending=False)
)
companies_multiple_roles.head(10)
```

Company Analysis

```python
# Top hiring companies
hiring_companies=df.groupby('company')['job_title'].count().sort_values(ascending=False).reset_index(name='Job Hirings')
hiring_companies.head(10)
```

```python
plt.Figure(figsize=(20,12))
top10=hiring_companies.head(10)
sns.lineplot(data=top10,x='Job Hirings',y='company')
plt.title('Top 10 Job Hiring Company')
```

```python
remote_jobs = df[df['job_type'] == 'Remote']
remote_jobs['company'].value_counts()
remote_jobs['company'].value_counts().head(10).reset_index(name='Remote Openings')
```

```python
# Companies hiring across multiple countries
df.columns
companies = (
    df.groupby('company')['search_country']
      .nunique()
      .reset_index(name='Count of Countries')
)
companies = companies[companies['Count of Countries'] > 1]
companies.sort_values(by='Count of Countries', ascending=False).head(10)
```

```python
(
    df[['company', 'search_country']]
      .drop_duplicates()
      ['company']
      .value_counts()
      .loc[lambda x: x > 1]
).reset_index().head(10)
```

```python
# Companies hiring different experience levels
pivot = pd.pivot_table(
    data=df,
    values='job level',
    index='company',
    aggfunc='nunique'
)

pivot[pivot['job level'] > 1].sort_values(
    by='job level',
    ascending=False
).head(10)
```

```python
# Companies posting highest variety of jobs
pivot = pd.pivot_table(
    data=df,
    values='job_title',
    index='company',
    aggfunc='nunique'
)

pivot.sort_values(by='job_title', ascending=False).head(10)
```

```python
top10=pivot.head(10)
plt.figure(figsize=(12, 6))
sns.barplot(data=top10, x='job_title', y='company')

plt.title('Top 10 Companies Posting the Highest Variety of Jobs')
plt.xlabel('Number of Unique Job Titles')
plt.ylabel('Company')
plt.tight_layout()
plt.show()
```
