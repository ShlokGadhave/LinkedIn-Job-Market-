Handle Missing Values


```python
df.isnull().sum()
```

```python
df['job_skills'].fillna('Not Specified', inplace=True)
df['job_summary'].fillna('Not Available', inplace=True)
df.to_csv("Jobs.csv", index=False)
df.head(3)
```

Check Duplicates Rows

```python
duplicate_rows=df.duplicated().sum()
print(f"Number of Duplicated Rows : {duplicate_rows}")
```

Feature Engineering


```python
df.head(1)
df['first_seen']=pd.to_datetime(df['first_seen'])
df['Year']=df['first_seen'].dt.year
df['Month']=df['first_seen'].dt.month
df['Day']=df['first_seen'].dt.day
df['Month_Name']=df['first_seen'].dt.month_name()
df[['first_seen', 'Year', 'Month', 'Day','Month_Name']].head()
df.to_csv("Jobs.csv", index=False)
```

```python
df.head(3)
```

```python
df['Number_of_Skills']=df['job_skills'].str.split(',').str.len()
df['Number_of_Skills'].head(10)

```

```python
df['Remote_Flag']=df['job_type'].map({'Onsite':0,'Remote':1,'Hybrid':2})
df['Remote_Flag'].head(5)
df.to_csv('Jobs.csv',index=False)
```
